```python
# ensure the visualization renders properly across VSCode, Jupyter Book, etc.
import plotly.io as pio
pio.renderers.default = "notebook_connected+plotly_mimetype"
```

```python
import plotly.express as px

df = px.data.tips()
df.head()
```

```python
fig = px.scatter(
    df,
    x="total_bill",
    y="tip",
    trendline="ols",
    title="Total bill vs. tip",
)
fig.show()
```
